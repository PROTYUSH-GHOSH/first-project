```python
# Step 3: Train YOLOv8 on GPU (Runs 50 epochs in ~4 minutes on Colab GPU!)
from ultralytics import YOLO

# Start with pretrained road damage or yolov8s
model = YOLO('yolov8s.pt')

results = model.train(
    data='dataset/data.yaml', # Or your uploaded data.yaml
    epochs=30,
    imgsz=640,
    batch=16,
    device=0
)
```
